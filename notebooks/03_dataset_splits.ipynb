# 03 — Dataset Splits

Create stratified train / val / test splits and audit for leakage.

```python
import sys; sys.path.insert(0, '..')
from src.data_utils import load_labels, make_splits, save_splits
import pandas as pd

df = load_labels()
train_df, val_df, test_df = make_splits(df)
print(f'train={len(train_df)}, val={len(val_df)}, test={len(test_df)}')
```

```python
# Class distribution per split
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1,3, figsize=(14,4), sharey=True)
for ax, (split_df, name) in zip(axes, [(train_df,'train'),(val_df,'val'),(test_df,'test')]):
    split_df['label'].value_counts().sort_index().plot(kind='bar', ax=ax, color='steelblue')
    ax.set_title(f'{name} (n={len(split_df)})')
plt.tight_layout(); plt.savefig('../outputs/figures/split_distribution.png', dpi=150); plt.show()
```

```python
# Leakage audit — no filename should appear in more than one split
train_set = set(train_df['filename'])
val_set   = set(val_df['filename'])
test_set  = set(test_df['filename'])
assert not (train_set & val_set),  'LEAK: train ∩ val'
assert not (train_set & test_set), 'LEAK: train ∩ test'
assert not (val_set   & test_set), 'LEAK: val ∩ test'
print('✓ No leakage detected.')
```

```python
save_splits(train_df, val_df, test_df)
```
