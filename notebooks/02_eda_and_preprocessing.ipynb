# 02 — EDA & Preprocessing

Exploratory data analysis, image QC, resize, and normalisation.

```python
import sys; sys.path.insert(0, '..')
from src.data_utils import load_labels, CLASSES
import pandas as pd, numpy as np, matplotlib.pyplot as plt

df = load_labels()
print(df.shape)
df.head()
```

```python
# Class distribution
ax = df['label'].value_counts().reindex(CLASSES).plot(kind='bar', color='steelblue', figsize=(8,4))
ax.set(title='Class distribution', xlabel='Style', ylabel='Count')
for p in ax.patches:
    ax.annotate(str(int(p.get_height())), (p.get_x()+p.get_width()/2, p.get_height()+1), ha='center')
plt.tight_layout(); plt.savefig('../outputs/figures/class_distribution.png', dpi=150); plt.show()
```

```python
# Sample images per class
from pathlib import Path
from PIL import Image
import random

IMG_ROOT = Path('../Dataset/images')
fig, axes = plt.subplots(len(CLASSES), 5, figsize=(15, 3*len(CLASSES)))
for r, cls in enumerate(CLASSES):
    cls_files = df[df.label==cls]['filename'].tolist()
    sample = random.sample(cls_files, min(5, len(cls_files)))
    for c, fname in enumerate(sample):
        p = IMG_ROOT / Path(fname).name
        if p.exists():
            axes[r,c].imshow(Image.open(p))
        axes[r,c].axis('off')
        if c==0: axes[r,c].set_title(cls, fontsize=10)
plt.tight_layout(); plt.savefig('../outputs/figures/sample_grid.png', dpi=100); plt.show()
```

```python
# Image size distribution
import warnings
sizes = []
for fname in df['filename']:
    p = IMG_ROOT / Path(fname).name
    if p.exists():
        with warnings.catch_warnings():
            warnings.simplefilter('ignore')
            try: sizes.append(Image.open(p).size)
            except: pass
widths, heights = zip(*sizes) if sizes else ([],[]) 
plt.figure(figsize=(8,3))
plt.subplot(1,2,1); plt.hist(widths, bins=30, color='coral'); plt.title('Width distribution')
plt.subplot(1,2,2); plt.hist(heights, bins=30, color='steelblue'); plt.title('Height distribution')
plt.tight_layout(); plt.show()
```
